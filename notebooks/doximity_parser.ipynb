## Imports

```python
try: 
    from BeautifulSoup import BeautifulSoup
except ImportError:
    from bs4 import BeautifulSoup
```

```python
fn = '../logs/54d02a5326f7b844d1e0bb7e80257f7f.html'

with open(fn,'rb') as infile:
    html = infile.read()
```

```python
parsed_html = BeautifulSoup(html)
```

## Extract basic info

```python
first_name = parsed_html.body.find('span', attrs={'class':'user-name-first'}).text
middle_name = parsed_html.body.find('span', attrs={'class':'user-name-middle'}).text
maiden_name = parsed_html.body.find('span', attrs={'class':'user-name-maiden'}).text
last_name = parsed_html.body.find('span', attrs={'class':'user-name-last'}).text
creds = parsed_html.body.find('span', attrs={'class':'user-name-credentials'}).text
profile_title = parsed_html.body.find('a', attrs={'class':'profile-head-subtitle'}).text

city = parsed_html.body.find('span', attrs={'itemprop':'addressLocality'}).text
state = parsed_html.body.find('span', attrs={'itemprop':'addressRegion'}).text

job_title = parsed_html.body.find('p', attrs={'class':'user-job-title'}).text

office_addr = parsed_html.body.find('span', attrs={'class':'black profile-contact-labels-wrap'}).text

div_phone_elem = parsed_html.body.find('div', attrs={'class':'profile-contact-labels-wrap'})
office_phone = div_phone_elem.find('span', attrs={'class':'black'}).text

div_fax_elem = parsed_html.body.find('div', attrs={'class':'office-info-fax'})
office_fax = div_fax_elem.find('span', attrs={'class':'black'}).text
```

## Extract education

```python
edu_section_elem = parsed_html.body.find('section', attrs={'class':'education-info'})
educations_items = edu_section_elem.find_all('li', attrs={'itemprop':'alumniOf'})

educations = []

for li_item in educations_items:
    university_name = li_item.find('span', attrs={'class':'black'}).text
    job_pos = li_item.find('span', attrs={'class':'br'}).text
    educations.append({"university_name": university_name, "job_pos": job_pos})
```

```python
educations
```

## Extract Graph

```python
similar_graph = []

similar_items = parsed_html.body.find_all('li', attrs={'class':'sidebar-pymk-list-item'})

for li_item in similar_items:
    similar_item_link = li_item.find('a')['href']
    similar_item_name = li_item.find('p', attrs={'class':'similar-profile-name'}).text
    span_items = li_item.find_all('span')
    
    attributes = []
    for span_item in span_items:
        if span_item.text:
            attributes.append(span_item.text)
    
    similar_graph.append({"link": similar_item_link, "name": similar_item_name, "field": attributes[0], "loc":  attributes[1]})
    
```
